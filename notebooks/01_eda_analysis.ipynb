```python
"""
Financial News EDA - Task 1
Includes: descriptive statistics, publisher analysis, time series, topic modeling
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import nltk
from nltk.corpus import stopwords

# Download NLTK data
nltk.download('stopwords', quiet=True)

print("="*60)
print("FINANCIAL NEWS EDA - TASK 1")
print("="*60)

# Create sample data (replace with real dataset)
np.random.seed(42)

headlines = [
    "Fed raises interest rates by 25 basis points",
    "Apple beats earnings expectations",
    "Oil prices surge on supply concerns",
    "Tesla stock drops after recall",
    "FDA approves new cancer drug",
    "Job market shows unexpected strength",
    "Inflation data lower than expected",
    "Fed signals potential rate cuts",
    "Bank of America price target raised",
    "Microsoft acquires AI startup"
]

publishers = ['Reuters', 'Bloomberg', 'WSJ', 'CNBC', 'Financial Times']
dates = pd.date_range(end=datetime.now(), periods=200, freq='H')

df = pd.DataFrame({
    'headline': np.random.choice(headlines, 200),
    'publisher': np.random.choice(publishers, 200),
    'timestamp': dates
})

print(f"\n📊 Dataset: {len(df)} articles")
print(f"📅 Date range: {df['timestamp'].min()} to {df['timestamp'].max()}")

# 1. Descriptive Statistics
df['char_count'] = df['headline'].str.len()
df['word_count'] = df['headline'].str.split().str.len()

print("\n" + "="*60)
print("📝 HEADLINE STATISTICS")
print("="*60)
print(f"Average characters: {df['char_count'].mean():.1f}")
print(f"Average words: {df['word_count'].mean():.1f}")
print(f"Median words: {df['word_count'].median():.0f}")

# 2. Publisher Analysis
print("\n" + "="*60)
print("📰 PUBLISHER ANALYSIS")
print("="*60)
publisher_counts = df['publisher'].value_counts()
for pub, count in publisher_counts.items():
    print(f"{pub}: {count} articles ({count/len(df)*100:.1f}%)")

# 3. Time Series Analysis
df['hour'] = df['timestamp'].dt.hour
peak_hour = df['hour'].mode()[0]
print("\n" + "="*60)
print("⏰ TIME SERIES INSIGHTS")
print("="*60)
print(f"Peak publishing hour: {peak_hour}:00")

# 4. Topic Modeling
stop_words = set(stopwords.words('english'))
tfidf = TfidfVectorizer(max_features=10, stop_words=list(stop_words))
tfidf_matrix = tfidf.fit_transform(df['headline'])
keywords = tfidf.get_feature_names_out()

print("\n" + "="*60)
print("🔑 TOP KEYWORDS")
print("="*60)
for keyword in keywords[:10]:
    print(f"- {keyword}")

print("\n✅ EDA Complete!")
print("="*60)
```
